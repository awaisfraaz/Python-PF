```python
#to do list 
tasks =["car wash","buy groceries","pay bills","call mom"]

tasks.append("clean the house")
print(tasks)
tasks.remove("buy groceries")
print(tasks)
if "pay bills" in tasks:
    print("You have to pay bills today")
print("The Remaining tasks are:")
tasks.sort()
for i, task in enumerate(tasks):
    print(f"{i+1}-{task}")
```

```python
#Grades 

Grades = [90, 85, 78, 92, 88]
Grades.sort()
max_grade = max(Grades)
min_grade = min(Grades)
avg_grade = sum(Grades) / len(Grades)
print(f"Max grade: {max_grade}")    
print(f"Min grade: {min_grade}")
print(f"Average grade: {avg_grade:.2f}")
# Students
```

```python
#Inventory
inventory = ["apple, banana", "orange", "grape"]
inventory.append("kiwi")
print("Updated inventory:", inventory)
if  "banana" in inventory:
    print("Banana is available in the inventory.")
else:   
    print("Banana is not available in the inventory.")

print("The avaliable inventory is:")
for i, item in enumerate(inventory):
    print(f"{i+1}-{item}")
```
