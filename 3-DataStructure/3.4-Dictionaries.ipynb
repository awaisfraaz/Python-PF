```python
empty_dict = {}
print(type(empty_dict))
empty_dict = dict()
print(type(empty_dict))
```

```python
info_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(info_dict)
#key is unique
print(info_dict['name'])
print(info_dict['age'])
```

```python
#Get function 
print(info_dict.get('name'))
print(info_dict.get('country'))  # Default value if key doesn't exist
```

```python
#Dictionary are mutable
info_dict['age'] = 31
print(info_dict)
info_dict['country'] = 'USA'  # Adding a new key-value pair
print(info_dict)
```

```python
del info_dict['country']  # Deleting a key-value pair
print(info_dict)
```

```python
values= info_dict.values()  # Get all values
print(values)
keys = info_dict.keys()  # Get all keys
print(keys)
```

```python
items = info_dict.items()  # Get all key-value pairs
print(items)    #they are tuples
```

```python
#shallow copy
new_dict = info_dict.copy()
print(new_dict)
```

```python
new_dict=info_dict.copy()
info_dict['name'] = 'Bob'
print(new_dict) 
print(info_dict)  # This creates a reference, not a copy
```

```python
#iterating through a dictionary
for key in info_dict:
    print(f"Key: {key}, Value: {info_dict[key]}")
```

```python
 #Netsed dictionary 
 student_info = {
    'student1': {'name': 'Alice', 'age': 20, 'grades': [85, 90, 95]},
    'student2': {'name': 'Bob', 'age': 22, 'grades': [88, 92, 94]}
}
print(student_info['student1']['name'])  # Accessing nested dictionary
print(student_info['student2']['grades'])  # Accessing nested list
print(student_info['student2']['grades'][2])  # Accessing nested list
```

```python
#iterating through a  nested dictionary
for student, details in student_info.items():
    print(f"Student: {student}")
    for key, value in details.items():
        print(f"  {key}: {value}")
        
```

```python
#Dictionary comprehension
squared_numbers = {x: x**2 for x in range(1, 6)}
print(squared_numbers)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
#conditional dictionary comprehension
even_squares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(even_squares)  # {2: 4, 4: 16, 6: 36, 8: 64, 10: 100}
```

```python
numbers = [1, 2,2, 3,3,3, 4, 5,5,5,5,5]
frequency_dict={}

for number in numbers:
    if number in frequency_dict:
        frequency_dict[number] += 1
    else:
        frequency_dict[number] = 1
print(frequency_dict)
```

```python
#merge two dictionaries
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}

merged_dict = {**dict1, **dict2}
print(merged_dict)
```
