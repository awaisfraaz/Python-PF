```python
#packing and unpacking
packed_tuple = (1, "Hello", 3.14)
print(packed_tuple)
a, b, c = packed_tuple
print(a)
print(b)
print(c)
```

```python
#Nested tuples 
nested_tuple = (1, (2, 3), (4, 5, 6), "Hello")
print(nested_tuple)
print(nested_tuple[1][1])  # Accessing the nested tuple (2, 3)
```
