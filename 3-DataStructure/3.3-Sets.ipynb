```python
#Matematical operations on sets
set1={1,2,3,4,5}
set2={4,5,6,7,8}
set3=set1.union(set2)  #Union
print(set3)
ser4t5=set1.intersection(set2)  #Intersection
```

```python
#Set difference 
set5=set1.difference(set2)  #Elements in set1 but not in set2
print(set5)
```

```python
set1.intersection_update(set2)  #Update set1 with intersection
print(set1)  #set1 is now {4, 5}
```

```python
#Set symmetric difference
set6=set1.symmetric_difference(set2)  #Elements in either set1 or set2 but not both
print(set6)  #{1, 2, 3, 6, 7, 8}

```

```python
#set methods 
set1={1, 2, 3, 4, 5}
set2={4, 5, 6, 7, 8}
print(set1.issubset(set2))  #False, set1 is not a subset of set2
print(set2.issubset(set1))  #False, set2 is not a subset of set1
print(set1.issuperset(set2))  #False, set1 is not a superset of set2
print(set2.issuperset(set1))  #False, set2 is not a superset of set1
```
