```python
number_dict={x: x**2 for x in range(11)}
print(number_dict)
```

```python
print("The square of 5 is:", number_dict[5])
```

```python
# Adding a new key-value pair
number_dict[11] = 121

del number_dict[1]  # Removing the key-value pair for 3
print("Updated dictionary:", number_dict)
```

```python
dict1={x: x**2 for x in range(5)}
print(dict1)
```

```python
dict2={x: x**2 for x in range(6,11)}
print(dict2)
# Merging two dictionaries
merged_dict = {**dict1, **dict2}    
print("Merged dictionary:", merged_dict)
```

```python
#Create a dictionary where the keys are the first 5 positive integers and the values are tuples containing the key and its square. Print the dictionary.
tuple_dict = {x: (x, x**2) for x in range(1, 6)}
print("Dictionary with tuples as values:", tuple_dict)
```

```python
#Create a dictionary with the first 10 positive integers as keys and their squares as values. Create a new dictionary containing only the key-value pairs where the key is even. Print the new dictionary.

even_squares_dict = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print("Dictionary with even keys:", even_squares_dict)  
```

```python
#Create a dictionary representing a book with keys 'title', 'author', 'year', and 'genre'. Convert the dictionary to a JSON string and print it.
import json
book_dict = {
    "title": "To Kill a Mockingbird",
    "author": "Harper Lee",
    "year": 1960,
    "genre": "Fiction"
}
book_json = json.dumps(book_dict)
print("Book JSON:", book_json)
print("Type of book_json:", type(book_json))

```
