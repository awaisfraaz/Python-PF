```python
#reading a whole file 
with open('data.txt', 'r') as file:
    data = file.read()
    print(data)
```

```python
#read line by line
with open('data.txt', 'r') as file:
    for line in file:
        print(line.strip())
        #strip() removes leading and trailing whitespace characters
        
```

```python
#writing to a file
with open('output.txt', 'w') as file:
    file.write("This is a new line in the file.\n")
    file.write("Another line added to the file.\n")
```

```python
#write a file without overwriting
with open('output.txt', 'a') as file:
    file.write("Appending a new line to the existing file.\n")
    file.write("Yet another line added to the file.\n")
```

```python
#writing the file in binary mode
data=b'\x01\x02\x03\x04\x05'
with open('output.bin', 'wb') as file:
    file.write(data)
#reading the file in binary mode
with open('output.bin', 'rb') as file:
    binary_data = file.read()
    print(binary_data)  # Output: b'\x01\x02\x03\x04\x05'
```

```python
#reads the file lines characters and words 
with open('data.txt', 'r') as file:
    lines = file.readlines()  # Read all lines into a list
    print("Lines:", lines)

    for line in lines:
        words = line.split()  # Split each line into words
        print("Words in line:", words)

        characters = list(line.strip())  # Convert each line to a list of characters
        print("Characters in line:", characters)
```

```python
#writing and then reading 
with open('data1.txt', 'w+') as file:
    file.write("This is a sample text file.\n")
    file.write("It contains multiple lines of text.\n")
        
    file.seek(0)  # Move the file pointer to the beginning

    content = file.read()  # Read the content of the file
    print(content)
```
