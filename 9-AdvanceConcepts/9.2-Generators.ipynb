```python
def mygenerator():
    yield 1
    yield 2
    yield 3
```

```python
gen=mygenerator()
gen
```

```python
for i in gen:
    print(i)
    
# here we are not creating the copy just passing values
```

```python
#practical example reading large file

def mygenerator(filename):
    with open(filename,'r') as f:
        for line in f:
            yield line
```

```python
filename='large.txt'
for line in mygenerator(filename):
    print(line.strip())
```
