```python
import array 
arr=array.array('i', [1, 2, 3, 4, 5])
print(arr)
# Output: array('i', [1, 2, 3, 4, 5])
# To create an array of unicode characters, use 'u' type code and pass a single string
arr1 = array.array('u', ['a', 'b', 'c', 'd', 'e'])
print(arr1)
```

```python
import random
print(random.randint(1, 10))  # Generates a random integer between 1 and 10
print(random.choice(arr))  # Randomly selects an element from the array
print(random.sample(arr, 3))  # Randomly selects 3 unique elements from the array
```

```python
import json
data = {"name": "jhon", "age": "10"}

json_string = json.dumps(data)  # Converts Python object to JSON string
print(json_string)  # Output: {"name": "jhon", "age": "
print(type(json_string))

json_data = json.loads(json_string)  # Converts JSON string to Python object
print(json_data)  # Output: {'name': 'jhon', 'age': '10'}
print(type(json_data))
```

```python
import csv
with open('data.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age'])
    writer.writerow(['Alice', 30])
    writer.writerow(['Bob', 25])
with open('data.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)  # Output: ['Name', 'Age'], ['Alice', '30'], ['Bob', '25']
```

```python
from datetime import datetime, timedelta
now = datetime.now()
print(now)  # Current date and time

yesterday = now - timedelta(days=1)
print(yesterday)  # Yesterday's date and time
```

```python
import time
time.sleep(2)  # Pauses the program for 2 seconds
print("Program resumed after 2 seconds")
print(time.time())  # Current time in seconds since the epoch
```

```python
#Regualr expressions
import re
pattern = r'\d+'  # Matches one or more digits
text = "There are 123 apples and 456 oranges."
matches = re.findall(pattern, text)
print(matches)  # Output: ['123', '456']
```
