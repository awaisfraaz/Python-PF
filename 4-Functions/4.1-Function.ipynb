```python
#even functino 
def even(n):
    if n % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")
even(10)

```

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b    

def mul(a, b):
    return a * b
def div(a, b):
    if b == 0:
        return "Division by zero is not allowed"
    return a / b
def mod(a, b):
    return a % b
x=add(10, 5)
print("Addition:", x)
```

```python
def welcome(name):
    print(f"Hello, {name}! Welcome to the program.")

welcome("Alice")
```

```python
#defalut perameter
def  welcome(name="User"):
    print(f"Hello, {name}! Welcome to the program.")

welcome()  # Calls with default parameter
welcome("Bob")  # Calls with a specific name
```

```python
#varible Length argument
#positional arguments
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

result = add_numbers(1, 2, 3, 4, 5)
print("Sum of numbers:", result)

```

```python
#keyword arguments dictionary
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=25, city="New York")
```

```python
#postional vs keyword arguments
def print_info(name, age, city):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")

print_info("Alice", 25, "New York")

print_info(age=25, name="Alice", city="New York")
```
