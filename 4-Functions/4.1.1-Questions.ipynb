```python
#temperature comversion 
def convert_temperature(temp, unit):
    if unit.lower() == 'c':
        # Convert Celsius to Fahrenheit
        return (temp * 9/5) + 32
    elif unit.lower() == 'f':
        # Convert Fahrenheit to Celsius
        return (temp - 32) * 5/9
    else:
        raise ValueError("Unit must be 'C' for Celsius or 'F' for Fahrenheit.")


# Example usage
celsius = 25
fahrenheit = convert_temperature(celsius, 'C')
print(f"{celsius} degrees Celsius is equal to {fahrenheit} degrees Fahrenheit.")
```

```python
#calculate the total cost in a shopping list
def total_cost(prices):
    total = 0
    for item in prices:
        total += item['price'] * item['quantity']
    return total

prices = [    {"item": "apple", "price": 0.5, "quantity": 4},
    {"item": "banana", "price": 0.3, "quantity": 6},
    {"item": "orange", "price": 0.8, "quantity": 3},
    {"item": "milk", "price": 1.2, "quantity": 2},
    {"item": "bread", "price": 1.0, "quantity": 1}
]
print(f"Total cost of shopping list: ${total_cost(prices):.2f}")
```

```python
#Check whether the string is palindrome or not
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]


# Example usage
is_palindrome("madam")  # Output: True
is_palindrome("hello")  # Output: False
```

```python
#calculating the factorial of a number
def factorial(n):
    if n== 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
# Example usage
print(factorial(5))  # Output: 120
```

```python
#Validate Email Address

def validator(email):
    if "@" in email and "." in email:
        return True
    else:
        return False

validator('jhon@gmail.com')
```
