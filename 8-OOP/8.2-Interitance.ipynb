```python
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    def showdata(self):
        print("Windows:", self.windows)
        print("Doors:", self.doors)
        print("Engine Type:", self.enginetype)  
    def drive(self):
        print("The per")
```

```python
#Inheritance
class Tesla(Car):
    
    def __init__(self,windows,doors,enginetype,is_autopilot):
        super().__init__(windows,doors,enginetype)
        self.is_autopilot =is_autopilot
    def selfdriving(self):
        print(f"Tesla supports self driveing{self.is_autopilot}")
```

```python
Tesla1=Tesla(4, 4, "Electric", True)
print(Tesla1.selfdriving())
```

```python
#Multiple  Inheritance 
class Animal:
    def __init__(self,name):
        self.name=name 
    def speak(self):
        print("The subclass must implement this functin")
```

```python
class Pet:
    def __init__(self,owner):
        self.owner=owner
```

```python
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)

    def speak(self):
        print(f"{self.name} say woof")
    
dog1=Dog("Alice","jhon")
dog1.speak() 





```
