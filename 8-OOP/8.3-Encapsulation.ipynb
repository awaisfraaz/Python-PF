```python
#public 
#private
#protected
class Person:
    def __init__(self, name):
        self.name = name  # public attribute
        self.__age = 0    # private attribute
    def get_age(self):
        return self.__age  # public method to access private attribute
    def input_age(self):
        self.__age = int(input("Enter age: "))  # method to set private attribute
  
p = Person("Alice")
print(p.name)          # Accessing public attribute
p.input_age()      # Setting private attribute through public method
print(p.get_age())     # Accessing public method
```
