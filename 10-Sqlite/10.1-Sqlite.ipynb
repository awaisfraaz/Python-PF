```python
import sqlite3
```

```python
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute(''' CREATE TABLE IF NOT EXISTS  Employee (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER NOT NULL,
    department TEXT NOT NULL
);''')
connection.commit()
```

```python
cursor.execute(''' SELECT * FROM Employee;''')
```

```python
#insert the data 
cursor.execute(''' INSERT INTO Employee (name, age, department) VALUES ('John Doe', 30, 'HR');''')
cursor.execute(''' INSERT INTO Employee (name, age, department) VALUES ('bob', 30, 'HR');''')
cursor.execute(''' INSERT INTO Employee (name, age, department) VALUES ('jane', 30, 'HR');''')
connection.commit()
```

```python
cursor.execute(''' SELECT * FROM Employee;''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#update the data
cursor.execute(''' UPDATE Employee SET age = 31 WHERE name = 'jane';''')
connection.commit()
```

```python
cursor.execute(''' SELECT * FROM Employee;''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#Delete the data
cursor.execute(''' DELETE FROM Employee WHERE name = 'bob';''')
connection.commit()
```

```python
#working with salesData base 
connection = sqlite3.connect('salesData.db')
cursor = connection.cursor()
cursor.execute(''' CREATE TABLE IF NOT EXISTS Sales (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    Date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER NOT NULL,
    Region TEXT NOT NULL
);''')
connection.commit() 
```

```python
sales_data = [
    ('2023-01-01', 'Product A', 100, 'North'),
    ('2023-01-02', 'Product B', 200, 'South'),
    ('2023-01-03', 'Product C', 150, 'East'),
    ('2023-01-04', 'Product D', 300, 'West'),
    ('2023-01-05', 'Product E', 250, 'North'),
    ('2023-01-06', 'Product F', 400, 'South'),
]
```

```python
cursor.executemany(''' INSERT INTO Sales (Date, product, sales, Region) VALUES (?, ?, ?, ?);''', sales_data)
connection.commit()
```

```python
cursor.execute(''' SELECT * FROM Sales;''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#close the connection 
connection.close()
```
