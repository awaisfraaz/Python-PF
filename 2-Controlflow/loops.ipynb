```python
#for loop
for i in range(5):
    print(i)
```

```python
for i in range(1, 10):
    print(i)
```

```python
#step size
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10, 1, -1):
    print(i)
```

```python
#print even number 
for i in range(0, 11, 2):
    print(i)
```

```python
str = "Hello guys"
for i in range(len(str)):
    print(str[i])
```

```python
for i in str:
    print(i)
```

```python
#while loop 
count=0

while count<10:
    print(count)
    count=count+1
```

```python
#Break statement
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
#continue statement
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
#Nested for loop
for i in range(3):
    for j in range(2):
        print(f"i={i} and j={j}")
```

```python
#Sum of N numbers
sum=0
count=1
n=int(input("Enter a number: "))
while count <= n:
    sum += count
    count += 1

print(f"The sum of first {n} numbers is: {sum}")
```

```python
#prime number printing
for num in range(2, 101):   
    if num>1:
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num)
            
    
```
